Debt-to-Income Ratio (DTI)

Debt-to-Income Ratio (DTI) represents the proportion of a person's income that is used to pay existing debts.

Formula:

DTI = Monthly Debt Payments / Monthly Income

Example:

Monthly Income = ₹50,000
Monthly Debt Payments = ₹15,000

DTI = 15,000 / 50,000
    = 0.30
    = 30%

Interpretation:

0.10 (10%) → Low debt burden
0.30 (30%) → Moderate debt burden
0.50 (50%) → High debt burden
0.70 (70%) → Very high debt burden
0.80+      → Extremely high debt burden

Generally, a lower DTI is considered better because it means the applicant has more of their income available after paying existing debts.

Why DTI is useful in LoanIQ:

DTI helps the model understand an applicant's existing financial burden. Two applicants may have the same income but very different debt obligations.

Applicant A:
Monthly Income = ₹60,000
Monthly Debt = ₹10,000

DTI = 10,000 / 60,000
    = 0.167 (16.7%)
Applicant B:
Monthly Income = ₹60,000
Monthly Debt = ₹40,000

DTI = 40,000 / 60,000
    = 0.667 (66.7%)

Both applicants earn the same amount, but Applicant B has a much higher existing debt burden.

```python
# Compare employment status with loan approval
sns.countplot(data=df, x="EmploymentStatus", hue="LoanApproved")

# Add a title
plt.title("Employment Status vs Loan Approval")

# Display the graph
plt.show()
```

What we observe

Most applicants in the dataset are Employed, so their bars are naturally much larger.

We cannot conclude from this graph that employed applicants are more likely to get approved just because the orange bar is larger. The dataset itself has many more employed applicants.

For the presentation:

"The count plot shows the number of approved and rejected applications across employment categories. Since the number of applicants differs significantly between categories, we need to compare approval rates rather than raw counts."

```python
# Calculate the average approval rate for each employment category
approval_rate = df.groupby("EmploymentStatus")["LoanApproved"].mean()

# Display the approval rate
approval_rate
```

```python
# Plot the approval rate for each employment category
approval_rate.plot(kind="bar")

# Add a title
plt.title("Loan Approval Rate by Employment Status")

# Label the axes
plt.xlabel("Employment Status")
plt.ylabel("Approval Rate")

# Display the graph
plt.show()
```

What we observe

The approval rates are:

Self-Employed: 27.84%
Employed: 24.00%
Unemployed: 18.19%

So in this dataset, Self-Employed applicants have the highest approval rate, while Unemployed applicants have the lowest.

For the presentation:

"Instead of comparing raw counts, we calculated the approval rate for each employment category. This gives a fairer comparison because the number of applicants in each category is different."
