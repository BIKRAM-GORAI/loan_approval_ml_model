Debt-to-Income Ratio (DTI)

Debt-to-Income Ratio (DTI) represents the proportion of a person's income that is used to pay existing debts.

Formula:

DTI = Monthly Debt Payments / Monthly Income

Example:

Monthly Income = ₹50,000
Monthly Debt Payments = ₹15,000

DTI = 15,000 / 50,000
    = 0.30
    = 30%

Interpretation:

0.10 (10%) → Low debt burden
0.30 (30%) → Moderate debt burden
0.50 (50%) → High debt burden
0.70 (70%) → Very high debt burden
0.80+      → Extremely high debt burden

Generally, a lower DTI is considered better because it means the applicant has more of their income available after paying existing debts.

Why DTI is useful in LoanIQ:

DTI helps the model understand an applicant's existing financial burden. Two applicants may have the same income but very different debt obligations.

Applicant A:
Monthly Income = ₹60,000
Monthly Debt = ₹10,000

DTI = 10,000 / 60,000
    = 0.167 (16.7%)
Applicant B:
Monthly Income = ₹60,000
Monthly Debt = ₹40,000

DTI = 40,000 / 60,000
    = 0.667 (66.7%)

Both applicants earn the same amount, but Applicant B has a much higher existing debt burden.

📊 What Is a Correlation Matrix?

A correlation matrix is basically a table that tells us:

"When one numerical variable changes, does another variable tend to change with it?"

For example, suppose:

Income ↑  →  Loan approval ↑

If this happens consistently in our data, income and loan approval have a positive correlation.

On the other hand:

Debt ↑  →  Loan approval ↓

would indicate a negative correlation.

Think of it like this:

    -1                  0                  +1
    │                   │                   │
    Strong              No                 Strong
    Negative         Linear              Positive
                       relationship
Positive correlation

Example:

    Income ↑
    
       ↓
       
    Loan approval tends to ↑

Correlation might be:

+0.60

The closer to +1, the stronger the positive relationship.

Negative correlation

Example:

    Debt ↑
    
       ↓
       
    Loan approval tends to ↓

Correlation might be:

-0.40

The closer to -1, the stronger the negative relationship.

Near zero

    Age ↑
    
       ↓
       
    No consistent change in approval

Correlation might be:

0.01

This means there is almost no linear relationship.

🔴 Red

Means positive correlation.

Example:

Very strong positive relationship.

🔵 Blue

                Selecting Required Columns
What this does:

X → contains the information used by the model.
y → contains the answer the model needs to learn: 0 = Rejected, 1 = Approved.

```python
# Select the 15 features we will use for prediction
features = [
    'Age',
    'AnnualIncome',
    'CreditScore',
    'EmploymentStatus',
    'EducationLevel',
    'LoanAmount',
    'LoanDuration',
    'MaritalStatus',
    'NumberOfDependents',
    'HomeOwnershipStatus',
    'MonthlyDebtPayments',
    'DebtToIncomeRatio',
    'BankruptcyHistory',
    'PreviousLoanDefaults',
    'PaymentHistory'
]

# Create input data
X = df[features]

# Create target data
y = df['LoanApproved']

# Display the selected data
X.head()
```

```python
# Check the number of rows and columns
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Select the features we will use for prediction
selected_columns = [
    'Age',
    'AnnualIncome',
    'CreditScore',
    'EmploymentStatus',
    'EducationLevel',
    'LoanAmount',
    'LoanDuration',
    'MaritalStatus',
    'NumberOfDependents',
    'HomeOwnershipStatus',
    'MonthlyDebtPayments',
    'DebtToIncomeRatio',
    'BankruptcyHistory',
    'PreviousLoanDefaults',
    'PaymentHistory',
    'LoanApproved'
]

# Create a new DataFrame with only these columns
loan_data = df[selected_columns]

# Display the first 5 rows
loan_data.head()
```

What this does
selected_columns → stores the columns we want.
df[selected_columns] → creates a smaller dataset using only those columns.
LoanApproved → remains our target/output.
The other 15 columns → become our input features.

Separate Features and Target

```python
# X contains the input features
X = loan_data.drop('LoanApproved', axis=1)

# y contains the target/output
y = loan_data['LoanApproved']

# Display the shapes
print("X shape:", X.shape)
print("y shape:", y.shape)
```

So:

    X → 20,000 applicants × 15 features
    y → 20,000 loan approval results

```python
# Split Training and Testing Data

from sklearn.model_selection import train_test_split

# Split the data into 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display the sizes of the datasets
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
# Identify Categorical and Numerical Features

# Select columns containing text values
categorical_features = X.select_dtypes(include=['str']).columns.tolist()

# Select columns containing numerical values
numerical_features = X.select_dtypes(exclude=['str']).columns.tolist()

# Display both groups
print("Categorical Features:")
print(categorical_features)

print("\nNumerical Features:")
print(numerical_features)
```

                        One-Hot Encoding
EmploymentStatus

    Employed
    Self-Employed
    Unemployed      

becomes something like:

    Employed | Self-Employed | Unemployed
        1    |       0       |     0
        0    |       1       |     0
        0    |       0       |     1

```python
# One-Hot Encoding and Preprocessing

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Create the preprocessing transformer
preprocessor = ColumnTransformer(
    transformers=[
        ('categorical', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ],
    remainder='passthrough'
)
```

```python
# Apply Preprocessing

# Fit the preprocessing rules using training data
X_train_processed = preprocessor.fit_transform(X_train)

# Apply the same rules to testing data
X_test_processed = preprocessor.transform(X_test)

# Display the shapes
print("Training data after preprocessing:", X_train_processed.shape)
print("Testing data after preprocessing:", X_test_processed.shape)
```

Perfect. 27 columns is expected.

We started with 15 input features, but the 4 categorical features expanded into multiple columns through One-Hot Encoding.

    15 original features
            ↓
    4 categorical features encoded
            ↓
    27 numerical features
            ↓
    Ready for ML models

                Train the First Machine Learning Model

```python
# Train Logistic Regression Model

from sklearn.linear_model import LogisticRegression

# Create the model
logistic_model = LogisticRegression(max_iter=1000)

# Train the model using training data
logistic_model.fit(X_train_processed, y_train)

# Make predictions on test data
y_pred_logistic = logistic_model.predict(X_test_processed)

# Display the predictions
print(y_pred_logistic[:10])
```

```python
# Scale Numerical Features

from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Scale the numerical columns
X_train_scaled = scaler.fit_transform(X_train_processed)
X_test_scaled = scaler.transform(X_test_processed)

# Display the shapes
print("Training data:", X_train_scaled.shape)
print("Testing data:", X_test_scaled.shape)
```

```python
# Retrain Logistic Regression Model

from sklearn.linear_model import LogisticRegression

# Create the model
logistic_model = LogisticRegression(max_iter=1000)

# Train the model
logistic_model.fit(X_train_scaled, y_train)

# Make predictions on the test data
y_pred_logistic = logistic_model.predict(X_test_scaled)

# Display the first 10 predictions
print(y_pred_logistic[:10])
```

Evaluate Logistic Regression
Now we need to check how well the model actually performed on the 20% test data.

We'll calculate:

    Accuracy
    Precision
    Recall
    F1 Score

```python
# Evaluate Logistic Regression Model

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred_logistic)
precision = precision_score(y_test, y_pred_logistic)
recall = recall_score(y_test, y_pred_logistic)
f1 = f1_score(y_test, y_pred_logistic)

# Display the results
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
```

                    Confusion Matrix

```python
# Create Confusion Matrix

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Calculate the confusion matrix
cm = confusion_matrix(y_test, y_pred_logistic)

# Display the confusion matrix
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Rejected", "Approved"]
)

disp.plot()

# Add a title
plt.title("Logistic Regression - Confusion Matrix")

# Display the graph
plt.show()
```

    "The confusion matrix helps us understand the types of errors made by the model. Out of the rejected applications, 2,854 were correctly identified, while 129 were incorrectly classified as approved. Among approved applications, 736 were correctly identified and 281 were incorrectly classified as rejected."

                    Train Decision Tree Model
Unlike Logistic Regression, a Decision Tree makes predictions using a series of if-else-like decisions.

For example:

                            Credit Score > 600?
                                   ↓
                                 Yes
                                   ↓
                            Income > 50,000?
                                   ↓
                                 Yes
                                   ↓
                               APPROVED

```python
# Train Decision Tree Model

from sklearn.tree import DecisionTreeClassifier

# Create the model
decision_tree = DecisionTreeClassifier(random_state=42)

# Train the model
decision_tree.fit(X_train_processed, y_train)

# Make predictions
y_pred_tree = decision_tree.predict(X_test_processed)

# Display the first 10 predictions
print(y_pred_tree[:10])
```

```python
# Evaluate Decision Tree Model

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Calculate evaluation metrics
accuracy_tree = accuracy_score(y_test, y_pred_tree)
precision_tree = precision_score(y_test, y_pred_tree)
recall_tree = recall_score(y_test, y_pred_tree)
f1_tree = f1_score(y_test, y_pred_tree)

# Display the results
print("Accuracy :", accuracy_tree)
print("Precision:", precision_tree)
print("Recall   :", recall_tree)
print("F1 Score :", f1_tree)
```



| Metric    | Logistic Regression | Decision Tree |

| Recall    |          **72.37%** |        67.85% |
| F1 Score  |          **78.21%** |        67.48% |



"The Decision Tree achieved 83.38% accuracy, which was lower than Logistic Regression. Its precision and F1-score were also lower, so based on the current results, Logistic Regression performs better."

                Confusion Matrix for Decision Tree

```python
# Decision Tree - Confusion Matrix

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Calculate the confusion matrix
cm_tree = confusion_matrix(y_test, y_pred_tree)

# Display the confusion matrix
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_tree,
    display_labels=["Rejected", "Approved"]
)

disp.plot()

# Add a title
plt.title("Decision Tree - Confusion Matrix")

# Display the graph
plt.show()
```

So:

    TN = 2645
    FP = 338
    FN = 327
    TP = 690

Compared with Logistic Regression, the Decision Tree makes more false-positive predictions (338 vs 129), which helps explain its lower precision.



                Train Random Forest Model

    Decision Tree 1 ──┐
    Decision Tree 2 ──┤
    Decision Tree 3 ──┤
    Decision Tree 4 ──┤──→ Final Prediction
           ...        │
    Decision Tree N ──┘

Random Forest combines many Decision Trees and combines their predictions to produce a more reliable result.

```python
# Train Random Forest Model

from sklearn.ensemble import RandomForestClassifier

# Create the model
random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model
random_forest.fit(X_train_processed, y_train)

# Make predictions
y_pred_forest = random_forest.predict(X_test_processed)

# Display the first 10 predictions
print(y_pred_forest[:10])
```

```python
# Evaluate Random Forest Model

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Calculate evaluation metrics
accuracy_forest = accuracy_score(y_test, y_pred_forest)
precision_forest = precision_score(y_test, y_pred_forest)
recall_forest = recall_score(y_test, y_pred_forest)
f1_forest = f1_score(y_test, y_pred_forest)

# Display the results
print("Accuracy :", accuracy_forest)
print("Precision:", precision_forest)
print("Recall   :", recall_forest)
print("F1 Score :", f1_forest)
```

                Create Prediction Function

Now we'll make the model usable for a new loan applicant.

The function will:

    Take applicant details.
    Apply the same encoding.
    Apply the saved scaler.
    Predict Approved / Rejected.

```python
# Save Processed Feature Names

feature_columns = preprocessor.get_feature_names_out()

joblib.dump(feature_columns, "loan_approval_features.pkl")

print(len(feature_columns))
print(feature_columns)
```

```python
# Save Preprocessing Objects

joblib.dump(preprocessor, "loan_approval_preprocessor.pkl")
joblib.dump(scaler, "loan_approval_scaler.pkl")

print("Preprocessor and scaler saved successfully.")
```

```python
# Loan Approval Prediction Function

def predict_loan(applicant):

    applicant_df = pd.DataFrame([applicant])

    processed = preprocessor.transform(applicant_df)

    scaled = scaler.transform(processed)

    prediction = final_model.predict(scaled)[0]

    return "Approved" if prediction == 1 else "Rejected"
```

```python
# Test Prediction

applicant = {
    "Age": 37,
    "AnnualIncome": 103264,
    "CreditScore": 594,
    "EmploymentStatus": "Employed",
    "EducationLevel": "Associate",
    "LoanAmount": 9184,
    "LoanDuration": 36,
    "MaritalStatus": "Married",
    "NumberOfDependents": 1,
    "HomeOwnershipStatus": "Mortgage",
    "MonthlyDebtPayments": 274,
    "DebtToIncomeRatio": 0.078884,
    "BankruptcyHistory": 0,
    "PreviousLoanDefaults": 0,
    "PaymentHistory": 26
}

print("Loan Prediction:", predict_loan(applicant))
```
