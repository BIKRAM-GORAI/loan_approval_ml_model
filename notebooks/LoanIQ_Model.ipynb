## EDA   --- Exploratory Data Analysis

```python
# Display the distribution of applicant ages
sns.histplot(data=df, x="Age", bins=20)

# Add a title
plt.title("Distribution of Applicant Age")

# Display the graph
plt.show()
```

"We used a histogram for Age because Age is a numerical variable and we wanted to understand its distribution across the applicants."

```python
# Display the distribution of annual income
sns.histplot(data=df, x="AnnualIncome", bins=30)

# Add a title
plt.title("Distribution of Annual Income")

# Display the graph
plt.show()
```

"The histogram shows that annual income is right-skewed, with most applicants concentrated at lower income levels and fewer applicants having very high incomes."

```python
# Display the distribution of credit scores
sns.histplot(data=df, x="CreditScore", bins=20)

# Add a title
plt.title("Distribution of Credit Score")

# Display the graph
plt.show()
```

"We used a histogram because CreditScore is numerical. The graph shows that most applicants have credit scores concentrated around the middle range, with relatively few applicants at the extremes."

```python
# Display the distribution of loan amounts
sns.histplot(data=df, x="LoanAmount", bins=30)

# Add a title
plt.title("Distribution of Loan Amount")

# Display the graph
plt.show()
```

"The histogram shows that most applicants request relatively smaller loan amounts, while a smaller number request significantly larger amounts, producing a right-skewed distribution."

```python
# Display the distribution of debt-to-income ratio
sns.histplot(data=df, x="DebtToIncomeRatio", bins=20)

# Add a title
plt.title("Distribution of Debt-to-Income Ratio")

# Display the graph
plt.show()
```

"We used a histogram to understand the distribution of the Debt-to-Income Ratio. Most applicants have relatively lower ratios, while fewer applicants have higher ratios."

```python
# Compare credit scores for approved and rejected loans
sns.boxplot(data=df, x="LoanApproved", y="CreditScore")

# Add a title
plt.title("Credit Score vs Loan Approval")

# Label the x-axis
plt.xlabel("Loan Approved (0 = No, 1 = Yes)")

# Display the graph
plt.show()
```
